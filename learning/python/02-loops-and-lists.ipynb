```python
#List of temperature readings from a sensor
temperatures = [72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4]

print("All readings. ", temperatures)
print("First reading: ", temperatures[0])
print("Last reading: ", temperatures[-1])
print("Total readings: ", len(temperatures))
```

```python
#Loop through all temperature readings
temperatures = [72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4]

for temp in temperatures:
    print(f"Reading: {temp} °C")
```

```python
#Automatic alarm checking on all readings
temperatures = [72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4]
alarm_limit = 80.0

for temp in temperatures:
    if temp > alarm_limit:
        print(f"⚠️ ALARM: {temp} °C - Exceeds limit!")
    else:
        print(f"✅ OK: {temp} °C - Normal")
```

```python
#Summary statistics from sensor readings 
temperatures = [72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4]

highest = max(temperatures)
lowest = min (temperatures)
average = sum(temperatures)/len(temperatures)

print(f"Highest: {highest} °C")
print(f"Lowest: {lowest} °C")
print(f"Aveage: {average: .2f} °C")
```

```python
#Mini temperature monitoring system
machine_name = "Conveyor belt A"
temperatures = [72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4]
alarm_limit = 80.0
alarm_count = 0

print(f"=== {machine_name} - Temperature Report ===\n")

for temp in temperatures:
    if temp > alarm_limit:
        print(f"⚠️ ALARM: {temp} °C")
        alarm_count += 1
    else:
        print(f"✅ OK: {temp} °C")

print(f"\n--- Summary ---")
print(f"Total Reading: {len(temperatures)}")
print(f"Highest: {max(temperatures)}°C")
print(f"Lowest: {min(temperatures)}°C")
print(f"Average: {sum(temperatures)/len(temperatures): .2f}°C")
print(f"Total alarms: {alarm_count}")
```
