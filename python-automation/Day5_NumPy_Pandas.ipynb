```python
import numpy as np

#Numpy array of sensor readings
temperatures = np.array([72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4])

print("Readings:", temperatures)
print("Average:", np.mean(temperatures).round(2))
print("Highest", np.max(temperatures))
print("Lowest", np.min(temperatures))
print("Std. Deviation:", np.std(temperatures).round(2))
```

```python
#Math on entire array at once
temperatures_celsius = np.array([72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4])

#Convert all readings to fahrenheit in one line
temperatures_fahrenheit = (temperatures_celsius * 9/5) +32

print("celcius: ", temperatures_celsius)
print("fahrenheit: ", temperatures_fahrenheit.round(1))

#check which reading exceeds limit
alarm_limit = 80.0
alarms = temperatures_celsius > alarm_limit

print("\nAlarm flags:", alarms)
print("Alarm readings:", temperatures_celsius[alarms])
```

```python
import pandas as pd

#Create a dataframe - factory sensor data
data = {
    "Machine": ["Conveyor Belt A", "Conveyor Belt B", "Conveyor Belt C", "Packaging", "Welding"],
    "Temperature": [83.2, 72.1, 75.5, 91.3, 78.0],
    "Motor_speed": [900, 1200, 1150, 800, 1100],
    "Is_Running": [True, True, True, True, False]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Exploring the DataFrame
print("shape:", df.shape)
print("\ncolumn names:", df.columns.tolist())
print("\nTemperature column only:")
print(df["Temperature"])
print("\nStatistics summary:")
print(df.describe().round(2))
```

```python
# Filtering machines by condition
alarm_limit = 80.0

# Machines above alarm limit
alarms = df[df["Temperature"] > alarm_limit]
print("=== Machines in ALARM ===")
print(alarms[["Machine", "Temperature"]])

# Machines not running
stopped = df[df["Is_Running"] == False]
print("\n=== Stopped Machines ===")
print(stopped[["Machine","Motor_speed"]])

# Machines running normally
normal = df[(df["Temperature"]< alarm_limit) & (df["Is_Running"] == True)]
print("\n=== Normal Machines ===")
print(normal[["Machine","Temperature"]])
```

```python
# Save DataFrame to CSV file
df.to_csv("factory_data.csv", index=False)
print("✅ Data saved to factory_data.csv")

# Read it back
df_loaded = pd.read_csv("factory_data.csv")
print("\nLoaded from CSV:")
print(df_loaded)
```
