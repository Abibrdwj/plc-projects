```python
import matplotlib.pyplot as plt
import numpy as np

#Temperature readings over time
temperatures = [72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4]
readings = [1, 2, 3, 4, 5, 6, 7]

plt.plot(readings, temperatures)
plt.title("Temperature Over Time")
plt.xlabel("Reading Number")
plt.ylabel("Temperature(°C)")
plt.grid(True)
plt.show()
```

```python
# Temperature chart with alarm line
temperatures = [72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4]
readings = [1, 2, 3, 4, 5, 6, 7]
alarm_limit = 80.0

plt.figure(figsize=(10,5))
plt.plot(readings, temperatures, marker="o", color="blue", label="Temperature")
plt.axhline(y=alarm_limit, color="red", linestyle="--", label="Alarm Limit")
plt.fill_between(readings, temperatures, alarm_limit,
                where=np.array(temperatures) > alarm_limit,
                color="red", alpha=0.3, label="Alarm Zone",
                interpolate=True)

plt.title("Temperature Monitoring - Converyor Belt A")
plt.xlabel("Reading Number")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
#Bar charts for comparing multiple machines
machines = ["Conveyor A", "Conveyor B", "Conveyor C", "Packaging", "Welding"]
temperatures = [83.2, 72.1, 75.5, 91.3, 78.0]
alarm_limit = 80.0

colors = ["red" if t > alarm_limit else "green" for t in temperatures]

plt.figure(figsize=(10,5))
plt.bar(machines, temperatures, color=colors)
plt.axhline(y=alarm_limit, color="red", linestyle="--", label="Alarm Limit")
plt.title("Factory Temperature Overview")
plt.xlabel("Machine")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(axis="y")
plt.show()
```

```python
# Multiple charts in one view
machines = ["Conveyor A", "Conveyor B", "Conveyor C", "Packaging", "Welding"]
temperatures = [83.2, 72.1, 75.5, 91.3, 78.0]
motor_speeds = [900, 1200, 1150, 800, 1100]
alarm_limit = 80.0

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Chart 1 - Temperature
colors = ["red" if t > alarm_limit else "green" for t in temperatures]
ax1.bar(machines, temperatures, color=colors)
ax1.axhline(y=alarm_limit, color="red", linestyle="--", label="Alarm Limit")
ax1.set_title("Temperature per Machine")
ax1.set_ylabel("Temperature (°C)")
ax1.legend()
ax1.grid(axis="y")

# Chart 2 - Motor Speed
ax2.bar(machines, motor_speeds, color="steelblue")
ax2.set_title("Motor Speed per Machine")
ax2.set_ylabel("Speed (RPM)")
ax2.grid(axis="y")

plt.tight_layout()
plt.show()
```

```python
# Save chart to file automatically
machines = ["Conveyor A", "Conveyor B", "Conveyor C", "Packaging", "Welding"]
temperatures = [83.2, 72.1, 75.5, 91.3, 78.0]
alarm_limit = 80.0

colors = ["red" if t > alarm_limit else "green" for t in temperatures]

plt.figure(figsize=(10, 5))
plt.bar(machines, temperatures, color=colors)
plt.axhline(y=alarm_limit, color="red", linestyle="--", label="Alarm Limit")
plt.title("Factory Temperature Report")
plt.xlabel("Machine")
plt.ylabel("Temperature (°C)")
plt.legend()
plt.grid(axis="y")
plt.savefig("factory_report.png", dpi=150, bbox_inches="tight")
plt.show()

print("✅ Chart saved to factory_report.png")
```
