```python
def check_temperature(temp, limit):
    if temp > limit:
        print(f"⚠️ ALARM: {temp} °C - Exceeds Limit!")
    else:
        print(f"✅ OK: {temp} °C - Normal")

check_temperature(75.3, 80.0)
check_temperature(85.1, 80.0)
check_temperature(68.9, 80.0)
```

```python
#Function that returns a value
def calculate_average(readings):
    average = sum(readings)/len(readings)
    return average

#using the returned value
temperatures = [72.1, 75.3, 80.5, 68.9, 77.2, 85.1, 71.4]
avg = calculate_average(temperatures)

print(f"Average Temperatures: {avg: .2f} °C")
```

```python
#Dictionary storing machine data
machine = {
    "name": "Conveyor Belt A",
    "temperature": 75.5,
    "motor_speed": 1200,
    "is_running": True,
    "alarm_limit": 80.0
}

print(machine["name"])
print(machine["temperature"])
print(machine["motor_speed"])
print(machine["is_running"])
print(machine["alarm_limit"])
```

```python
#updating machine data
machine = {
    "name": "Conveyor Belt A",
    "temperature": 75.5,
    "motor_speed": 1200
}

print(f"Before: {machine['temperature']} °C")

#update temperature and motor_speed
machine["temperature"] = 83.2
machine["motor_speed"] = 900

print(f"After: {machine['temperature']} °c")
print(f"New Motor Speed: {machine['motor_speed']} RPM")
```

```python
#complete machine monitor using functions + Dictionaries
machine = {
    "name": "Conveyor Belt A",
    "temperature": 83.2, 
    "motor_speed": 900,
    "is_running": True,
    "alarm_limit": 80.0
}
def check_machine(machine):
    print(f"\n=== {machine['name']} - status report ===")
    print(f"Running: {machine['is_running']}")
    print(f"Motor Speed: {machine['motor_speed']} RPM")
    print(f"Temperature: {machine['temperature']} °C")

    if machine["temperature"] > machine["alarm_limit"]:
        print(f"⚠️ ALARM: Temperature exceeds {machine['alarm_limit']} °C limit!")
    else:
        print(f"✅ Temperature Normal")

check_machine(machine)
```
