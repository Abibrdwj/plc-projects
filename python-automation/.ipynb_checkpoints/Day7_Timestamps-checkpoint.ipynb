```python
import datetime
def get_timestamp():
    return datetime.datetime.now().strftime("%y-%m-%d %H:%M:%S")

machines = [
    {"name": "Conveyor Belt A", "temperature" : 83.2, "alarm_limit":80.0},
    {"name":"Conveyor Belt B", "temperature": 72.1, "alarm_limit":80.0 },
    {"name": "Conveyor Belt C", "temperature": 75.5, "alarm_limit": 80.0},
]

with open("machine_log.txt", "w")as f:
    for machine in machines:
        timestamp = get_timestamp()
        status = "ALARM" if machine["temperature"] > machine["alarm_limit"] else "OK"
        line = f"[{timestamp}] {machine['name']} - {machine['temperature']} °C - {status}\n"
        print(line.strip())
        f.write(line)
print("\n✅ Log saved to machine_log.txt")
```

```python
import datetime

def get_timestamp():
    return datetime.datetime.now().strftime("%y-%m-%d %H:%M:%S")
    
def log_reading(machine_name, temperature, alarm_limit):
    status = "ALARM" if temperature > alarm_limit else "OK"
    timestamp = get_timestamp()
    line = f"[{timestamp}] {machine_name} - {temperature}°C - {status}\n"

    with open("machine_log.txt", "a")as f:
        f.write(line)
    
    print(line.strip())

log_reading("Conveyor Belt A", 83.2, 80.0)
```

```python
log_reading("Conveyor Belt A", 79.1, 80.0)
```

```python
log_reading("Conveyor Belt A", 85.5, 80.0)
```

```python
log_reading("Conveyor Belt A", 83.2, 80.0)
log_reading("Conveyor Belt A", 79.1, 80.0)
log_reading("Conveyor Belt A", 85.5, 80.0)
```

```python
import datetime
import csv

def get_timestamp():
    return datetime.datetime.now().strftime("%y-%m-%d %H:%M:%S")

machines = [
    {"name": "Conveyor Belt A", "temperature": 83.2, "alarm_limit": 80.0},
     {"name": "Conveyor Belt B", "temperature": 72.1, "alarm_limit": 80.0},
    {"name": "Conveyor Belt C", "temperature": 75.5, "alarm_limit": 80.0},
    {"name": "Packaging Unit", "temperature": 91.3, "alarm_limit": 85.0},
    {"name": "Welding Station", "temperature": 78.0, "alarm_limit": 80.0},
]

with open("factory_log.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Timestamp", "Machine", "Temperature", "Status"])

    for machine in machines:
        timestamp = get_timestamp()
        status = "ALARM" if machine["temperature"] > machine["alarm_limit"] else "OK"
        writer.writerow([timestamp, machine["name"], machine["temperature"], status])
        print(f"{timestamp} | {machine['name']} | {machine['temperature']} | {status}")
print("\n✅ CSV log saved to factory_log.csv")    
```

```python
#datetime.now() - Captures current time
#strftime() - Formats time into readable string
#Append mode "a" - Adds to log without overwriting
#csv.writer - Writes structured CSV logs
#Timestamped log function - Reusable logging for any machine
```
