```python
#Reading the machine data from a file
machine_log = open("machine_log.txt", "r")

content = machine_log.read()
machine_log.close()

print(content)
```

```python
#Better way - using "with" statement
with open("machine_log.txt", "r") as f:
    for line in f:
        print(line.strip())
```

```python
#Multiple machines stored as a list of dictionaries
machines = [
    {"name": "Conveyor Belt A", "Temperature": 83.2, "alarm_limit": 80.0},
    {"name": "Conveyor Belt B", "Temperature": 72.1, "alarm_limit": 80.0},
    {"name": "Conveyor Belt C", "Temperature": 72.5, "alarm_limit": 80.0},
    {"name": "Packaging unit", "Temperature": 91.3, "alarm_limit": 85.0},
    {"name": "Welding station", "Temperature": 78.0, "alarm_limit": 85.0},
]

for machine in machines:
    status = "⚠️ ALARM" if machine["Temperature"] > machine["alarm_limit"] else "✅OK"
    print(f"{machine['name']}: {machine['Temperature']} °C - {status}")
```

```python
#Full monitoring + logging system
machines = [
    {"name": "Conveyor Belt A", "temperature": 83.2, "alarm_limit": 80.0},
    {"name": "Conveyor Belt B", "temperature": 72.1, "alarm_limit": 80.0},
    {"name": "Conveyor Belt C", "temperature": 75.5, "alarm_limit": 80.0},
    {"name": "Packing Station", "temperature": 91.3, "alarm_limit": 85.0},
    {"name": "Welding Station", "temperature": 78.0, "alarm_limit": 80.0},
]

alarm_counts = 0

with open("factory_report.txt", "w") as f:
    f.write("=== Factory status Report ===\n\n")

    for machine in machines:
        if machine["temperature"] > machine["alarm_limit"]:
            status = "ALARM"
            alarm_counts += 1
        else: 
            status = "OK"
        
        line = f"{machine['name']}: {machine['temperature']} °C - {status}\n"
        print(line.strip())
        f.write(line)

    summary = f"\nTotal Machines: {len(machines)} | Total Alarms: {alarm_counts}"
    print(summary)
    f.write(summary)

print("\n✅ Report saved to factory_report.txt")
```
